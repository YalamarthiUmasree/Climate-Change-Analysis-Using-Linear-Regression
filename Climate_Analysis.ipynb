```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

print("Climate Change Analysis Project Started")
```

```python
data = {
    'Year': [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020],
    'AverageTemperature': [14.5,14.6,14.7,14.8,14.9,15.0,15.2,15.3,15.4,15.5,15.7]
}

df = pd.DataFrame(data)

df
```

## Train the Linear Regression Model

```python
X = df[['Year']]
y = df['AverageTemperature']

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)

print("Model Trained Successfully")
```

## Visualize the Results

```python
plt.figure(figsize=(8,5))

plt.scatter(X, y, label="Actual Temperature")
plt.plot(X, predictions, label="Regression Line")

plt.xlabel("Year")
plt.ylabel("Average Temperature")
plt.title("Climate Change Analysis Using Linear Regression")

plt.legend()

plt.show()
```
